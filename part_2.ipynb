### Split training data

```python
# We will use the same code we used for ex_3

def split_data(x, y, ratio, seed=1):
    """
    split the dataset based on the split ratio. If ratio is 0.8
    you will have 80% of your data set dedicated to training
    and the rest dedicated to testing. If ratio times the number of samples is not round
    you can use np.floor. Also check the documentation for np.random.permutation,
    it could be useful.

    Args:
        x: numpy array of shape (N,), N is the number of samples.
        y: numpy array of shape (N,).
        ratio: scalar in [0,1]
        seed: integer.

    Returns:
        x_tr: numpy array containing the train data.
        x_te: numpy array containing the test data.
        y_tr: numpy array containing the train labels.
        y_te: numpy array containing the test labels.

    >>> split_data(np.arange(13), np.arange(13), 0.8, 1)
    (array([ 2,  3,  4, 10,  1,  6,  0,  7, 12,  9]), array([ 8, 11,  5]), array([ 2,  3,  4, 10,  1,  6,  0,  7, 12,  9]), array([ 8, 11,  5]))
    """
    # set seed
    np.random.seed(seed)
    # ***************************************************
    # INSERT YOUR CODE HERE
    # split the data based on the given ratio: TODO
    # ***************************************************
    # Randomly shuffle the indices
    indices = np.random.permutation(len(x))

    # Number of training samples
    split_sample = int(np.floor(ratio * len(x)))

    # Split the shuffled indices
    train_indices = indices[:split_sample]
    test_indices = indices[split_sample:]

    # Use SAME indices for x and y
    x_tr = x[train_indices]
    x_te = x[test_indices]
    y_tr = y[train_indices]
    y_te = y[test_indices]

    return x_tr, x_te, y_tr, y_te
```

```python
train_X, test_X, train_y, test_y = split_data(x_train, y_train, 0.8, seed=2026)

print("Training data shape after split:", train_X.shape)
print("Test data shape after split:", test_X.shape)
```

```python
def print_label_distribution(y, name):
    values, counts = np.unique(y, return_counts=True)

    print(f"\n{name}")
    print("-" * 40)

    for value, count in zip(values, counts):
        percentage = 100 * count / len(y)
        print(
            f"Label {value}: {count} samples "
            f"({percentage:.2f}%)"
        )


print_label_distribution(train_y, "TRAINING SET")
print_label_distribution(test_y, "VALIDATION SET")
```

### Exploratory data analysis on the features

```python
# Missing-value analysis
num_samples, num_features = train_X.shape

print("Number of features:", num_features)

for j in range(num_features):
    missing = np.sum(np.isnan(train_X[:, j]))
    percentage = 100 * missing / num_samples

    if percentage > 0:
        print(
            f"Feature {j}: "
            f"{missing} missing values "
            f"({percentage:.2f}%)"
        )
```

```python
# Unique values per feature

for j in range(num_features):
    col = train_X[:, j]

    # Ignore NaNs
    valid = col[~np.isnan(col)]

    unique, counts = np.unique(valid, return_counts=True)

    print(f"\nFeature {j}")
    print("Number of unique values:", len(unique))

    if len(unique) <= 20:
        print("Values:", unique)
        print("Counts:", counts)
```

```python
# Identify constant features (features with only one unique value)
constant_features = []

for j in range(num_features):
    col = train_X[:, j]
    valid = col[~np.isnan(col)]

    if len(np.unique(valid)) <= 1:
        constant_features.append(j)

print("Constant features:", constant_features)
print("Number:", len(constant_features))
```

```python
missingness_results = []

for j in range(train_X.shape[1]):

    col = train_X[:, j]

    missing_mask = np.isnan(col)
    observed_mask = ~missing_mask

    n_missing = np.sum(missing_mask)
    n_observed = np.sum(observed_mask)

    # Need both groups for a meaningful comparison
    if n_missing > 0 and n_observed > 0:

        p_positive_missing = np.mean(
            train_y[missing_mask] == 1
        )

        p_positive_observed = np.mean(
            train_y[observed_mask] == 1
        )

        difference = abs(
            p_positive_missing - p_positive_observed
        )

        missing_fraction = n_missing / len(col)

        missingness_results.append(
            (
                j,
                missing_fraction,
                p_positive_missing,
                p_positive_observed,
                difference,
                n_missing,
                n_observed
            )
        )


# Sort by largest difference between missing and observed
missingness_results.sort(
    key=lambda x: x[4],
    reverse=True
)


print(
    "Feature | Missing % | P(+1 | missing) | "
    "P(+1 | observed) | Difference"
)
print("-" * 80)

for result in missingness_results:

    (
        feature,
        missing_fraction,
        p_missing,
        p_observed,
        difference,
        n_missing,
        n_observed
    ) = result

    print(
        f"{feature:7d} | "
        f"{100 * missing_fraction:8.2f}% | "
        f"{p_missing:15.4f} | "
        f"{p_observed:16.4f} | "
        f"{difference:10.4f}"
    )
```

```python
constant_features = [9, 11, 12, 18, 19, 22]

print(
    "Feature | Missing % | P(+1 | missing) | "
    "P(+1 | observed)"
)
print("-" * 70)

for j in constant_features:

    col = train_X[:, j]

    missing_mask = np.isnan(col)
    observed_mask = ~missing_mask

    missing_fraction = np.mean(missing_mask)

    if np.any(missing_mask):
        p_missing = np.mean(
            train_y[missing_mask] == 1
        )
    else:
        p_missing = np.nan

    if np.any(observed_mask):
        p_observed = np.mean(
            train_y[observed_mask] == 1
        )
    else:
        p_observed = np.nan

    print(
        f"{j:7d} | "
        f"{100 * missing_fraction:8.2f}% | "
        f"{p_missing:15.4f} | "
        f"{p_observed:16.4f}"
    )
```

```python
duplicate_pairs = []

num_features = train_X.shape[1]

for i in range(num_features):

    for j in range(i + 1, num_features):

        col_i = train_X[:, i]
        col_j = train_X[:, j]

        same_values = (
            (col_i == col_j) |
            (np.isnan(col_i) & np.isnan(col_j))
        )

        if np.all(same_values):
            duplicate_pairs.append((i, j))


print("\nExact duplicate feature pairs:")
print("-" * 40)

if len(duplicate_pairs) == 0:
    print("No exact duplicate features found.")
else:
    for i, j in duplicate_pairs:
        print(f"Feature {i} == Feature {j}")

print(
    "\nNumber of duplicate pairs:",
    len(duplicate_pairs)
)
```

```python
feature_summary = []

for j in range(train_X.shape[1]):

    col = train_X[:, j]

    missing_mask = np.isnan(col)
    valid = col[~missing_mask]

    n_missing = np.sum(missing_mask)
    missing_percentage = (
        100 * n_missing / len(col)
    )

    if len(valid) == 0:
        n_unique = 0
        most_common_fraction = np.nan
        min_value = np.nan
        max_value = np.nan

    else:
        unique, counts = np.unique(
            valid,
            return_counts=True
        )

        n_unique = len(unique)

        most_common_fraction = (
            np.max(counts) / len(valid)
        )

        min_value = np.min(valid)
        max_value = np.max(valid)

    feature_summary.append(
        (
            j,
            missing_percentage,
            n_unique,
            most_common_fraction,
            min_value,
            max_value
        )
    )


print(
    "Feature | Missing % | Unique | "
    "Dominant value % | Min | Max"
)
print("-" * 80)

for result in feature_summary:

    (
        feature,
        missing_percentage,
        n_unique,
        dominant_fraction,
        min_value,
        max_value
    ) = result

    print(
        f"{feature:7d} | "
        f"{missing_percentage:8.2f}% | "
        f"{n_unique:6d} | "
        f"{100 * dominant_fraction:15.2f}% | "
        f"{min_value:8.2f} | "
        f"{max_value:8.2f}"
    )
```

```python
thresholds = [0.50, 0.75, 0.90, 0.95, 0.99]

for threshold in thresholds:

    sparse_features = []

    for j in range(train_X.shape[1]):

        missing_fraction = np.mean(
            np.isnan(train_X[:, j])
        )

        if missing_fraction >= threshold:
            sparse_features.append(j)

    print(
        f"\nFeatures >= {threshold * 100:.0f}% missing:"
    )
    print(
        f"Count: {len(sparse_features)}"
    )
    print(sparse_features)
```

```python
dominant_results = []

for j in range(train_X.shape[1]):

    col = train_X[:, j]
    valid = col[~np.isnan(col)]

    if len(valid) == 0:
        continue

    values, counts = np.unique(
        valid,
        return_counts=True
    )

    largest_index = np.argmax(counts)

    dominant_value = values[largest_index]
    dominant_fraction = (
        counts[largest_index] / len(valid)
    )

    if dominant_fraction >= 0.95:

        dominant_results.append(
            (
                j,
                dominant_value,
                dominant_fraction,
                len(values),
                len(valid)
            )
        )


dominant_results.sort(
    key=lambda x: x[2],
    reverse=True
)


print(
    "Feature | Dominant value | Dominant % | "
    "Unique | Observed N"
)
print("-" * 70)

for (
    feature,
    value,
    fraction,
    n_unique,
    n_observed
) in dominant_results:

    print(
        f"{feature:7d} | "
        f"{value:14.2f} | "
        f"{100 * fraction:9.2f}% | "
        f"{n_unique:6d} | "
        f"{n_observed:10d}"
    )
```
